```python
alphabet = "абвгдеёжзиклмнопрстуфхцчшщъыьэюя .,"
characters = set(alphabet)
```

```python
file_to_analyze = input("Enter the path to file to analyze")
chars = open(file_to_analyze, "r", encoding="utf-8").read().lower()
```

```python


filtered_characters = "".join(filter(lambda char: char in alphabet, chars))
```

```python
from collections import Counter
import matplotlib.pyplot as plt

freq = Counter(filtered_characters)
```

```python
dictionary = {} # we need to reorder values for pretty histograms
for char in alphabet:
    dictionary[char] = freq[char]
```

```python
plt.figure(figsize=(10, 6))
plt.bar(dictionary.keys(), dictionary.values(), color='skyblue')
plt.xlabel('Characters')
plt.ylabel('Number of Occurrences')
plt.title('Character Frequency Histogram')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
