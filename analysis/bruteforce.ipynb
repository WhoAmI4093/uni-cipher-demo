```python
alphabet = "абвгдеёжзиклмнопрстуфхцчшщъыьэюя.,"
kw = ["был", "знаю", "этого"]
```

```python
file_path = input("Input file path")
chars = open(file_path, encoding="utf-8").read()[:1000]
```

```python
def try_decrypt_affine(alpha, beta, text):
    s = ""
    for char in text:
        idx = alphabet.index(char)
        resulting_idx = (alpha * ( idx - beta)) % len(alphabet)
        s += alphabet[resulting_idx]
    return s
```

```python
# Affine brute force
import math

for alpha in range(1, len(alphabet)):
    if math.gcd(alpha, len(alphabet)) != 1:
        continue

    for beta in range(0, len(alphabet)):
        text = try_decrypt_affine(alpha, beta, chars)
        # print("trying: ", text[:10])
        # print(list(map(lambda word: word in text ,kw)))
        if all(list(map(lambda word: word in text ,kw))):
            print(f"Inverse alpha: {alpha}; beta: {beta}")
```

```python
def try_decrypt_recursive_affine(alpha_1, beta_1, alpha_2, beta_2, text):
    s = ""
    for char in text:
        idx = alphabet.index(char)
        resulting_idx = (alpha_1 * ( idx - beta_1) ) % len(alphabet)
        # print(alphabet[resulting_idx])
        s += alphabet[resulting_idx]

        t_a = alpha_1
        t_b = beta_1

        alpha_1 = alpha_2
        beta_1 = beta_2

        alpha_2 = (t_a * alpha_2) % len(alphabet)
        beta_2 = (t_b + beta_2) % len(alphabet)
    
    return s
```

```python
# Recursive affine brute force
import math

for alpha_1 in range(1, len(alphabet)):
    if math.gcd(alpha_1, len(alphabet)) != 1:
        continue
    #print(alpha_1)
    for alpha_2 in range(1, len(alphabet)):
        if math.gcd(alpha_2, len(alphabet)) != 1:
            continue

        for beta_1 in range(len(alphabet)):
            for beta_2 in range(len(alphabet)):
                text = try_decrypt_recursive_affine(alpha_1, beta_1, alpha_2, beta_2, chars)
                if all(list(map(lambda word: word in text ,kw))):
                    print(f"Inverse alpha_1: {alpha_1}; beta_1: {beta_1}")
                    print(f"Inverse alpha_2: {alpha_2}; beta_2: {beta_2}")
```
